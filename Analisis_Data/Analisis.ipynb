```python
import pandas as pd

df = pd.read_csv("Data_Penjualan_Toko_Online.csv")
print(df.head())
```

```python
print(df.info())
print(df.isnull().sum())
# Jika ada nilai kosong
df = df.dropna()
```

```python
print("Statistik Deskriptif:")
print(df.describe())
print("\nProduk Terlaris:")
print(df.groupby("Produk")["Jumlah"].sum().sort_values(ascending=
False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Produk", y="Pendapatan", estimator=sum, ci=None)
plt.title("Total Pendapatan per Produk")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(data=df, x="Tanggal", y="Pendapatan", marker="o")
plt.title("Tren Pendapatan Mingguan")
plt.show()
```

```python
df.to_csv("hasil_analisis_penjualan.csv", index=False)
print("Data hasil analisis telah disimpan sebagai hasil_analisis_penjualan.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Data_Penjualan_Toko_Online.csv")
print("Data awal:")
print(df.head())
```

```python
df["Diskon"] = 0.10  # 10% dari harga untuk semua produk
```

```python
df["Pendapatan Bersih"] = df["Pendapatan"] - (df["Pendapatan"] * df["Diskon"])
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Produk", y="Pendapatan Bersih", estimator=sum, ci=None)
plt.title("Pendapatan Bersih per Produk (Setelah Diskon 10%)")
plt.xlabel("Produk")
plt.ylabel("Pendapatan Bersih")
plt.show()
```

```python
df.to_csv("analisis_diskon.csv", index=False)
print("✅ Data hasil analisis telah disimpan sebagai 'analisis_diskon.csv'")
```
